# Data Exploration for Custom TTS Voice Model

In this notebook, we will explore the dataset used for training the custom text-to-speech voice model. We will visualize audio features and perform necessary data analysis to understand the characteristics of the data before training the model.

```python
# Import necessary libraries
import os
import pandas as pd
import librosa
import matplotlib.pyplot as plt
import seaborn as sns

# Set the style for seaborn
sns.set(style='whitegrid')

# Define the path to the raw audio files
raw_audio_path = '../data/raw/'

# List all audio files in the raw data directory
audio_files = [f for f in os.listdir(raw_audio_path) if f.endswith('.wav')]
print(f'Number of audio files: {len(audio_files)}')

# Function to visualize audio waveform
def plot_waveform(audio_file):
    audio_path = os.path.join(raw_audio_path, audio_file)
    y, sr = librosa.load(audio_path)
    plt.figure(figsize=(12, 4))
    librosa.display.waveshow(y, sr=sr)
    plt.title(f'Waveform of {audio_file}')
    plt.xlabel('Time (s)')
    plt.ylabel('Amplitude')
    plt.show()

# Visualize the waveform of the first audio file
if audio_files:
    plot_waveform(audio_files[0])

# Function to extract and visualize audio features
def extract_features(audio_file):
    audio_path = os.path.join(raw_audio_path, audio_file)
    y, sr = librosa.load(audio_path)
    mfccs = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=13)
    return mfccs

# Extract features for the first audio file
if audio_files:
    mfccs = extract_features(audio_files[0])
    plt.figure(figsize=(12, 4))
    sns.heatmap(mfccs, cmap='viridis')
    plt.title('MFCCs of the first audio file')
    plt.xlabel('Frames')
    plt.ylabel('MFCC Coefficients')
    plt.show()

# Further analysis can be added here...
```
